```python
# T9: Support Vector Machine Classification
# Dataset: Most Followed Instagram Accounts Daily

# STEP 1: Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import GroupShuffleSplit
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# STEP 2: Load dataset

df = pd.read_csv('/content/daily.csv')

print("Dataset Shape:", df.shape)
print("\nFirst five records:")
print(df.head())

print("\nDataset Information:")
df.info()

# STEP 3: Data inspection

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nTarget Class Distribution:")
print(df['type'].value_counts())

# STEP 4: Remove duplicate records

df = df.drop_duplicates().copy()

# STEP 5: Convert date and extract features

df['date'] = pd.to_datetime(df['date'], errors='coerce')

df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day

# Remove records with invalid target values
df = df.dropna(subset=['type', 'username'])

# STEP 6: Separate input features and target

# Username and name are identifiers and are not used as predictors.
# also_in has too many missing values, so it is excluded.

X = df[
    ['followers', 'rank', 'year', 'month', 'day',
     'category', 'country']
].copy()

y = df['type'].astype(str)
groups = df['username'].copy()

print("\nInput Features:", X.columns.tolist())
print("Target Variable: type")

# STEP 7: Split into training and testing sets
# Keep each username in only one split to reduce data leakage.

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))

print("\nTraining Class Distribution:")
print(y_train.value_counts())

# STEP 8: Preprocessing

numeric_features = [
    'followers', 'rank', 'year', 'month', 'day'
]

categorical_features = [
    'category', 'country'
]

numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(
        strategy='constant',
        fill_value='Unknown'
    )),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('numeric', numeric_pipeline, numeric_features),
    ('categorical', categorical_pipeline, categorical_features)
])

# STEP 9: Build SVM model

svm_model = Pipeline([
    ('preprocessing', preprocessor),
    ('classifier', LinearSVC(
        C=1.0,
        class_weight='balanced',
        random_state=42,
        max_iter=10000
    ))
])

# STEP 10: Train the model

svm_model.fit(X_train, y_train)

print("\nSVM model trained successfully!")

# STEP 11: Make predictions

y_pred = svm_model.predict(X_test)

print("\nActual Values:")
print(y_test.head(10).tolist())

print("\nPredicted Values:")
print(y_pred[:10].tolist())

# STEP 12: Evaluate model performance

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(
    y_test, y_pred, average='weighted', zero_division=0
)
recall = recall_score(
    y_test, y_pred, average='weighted', zero_division=0
)
f1 = f1_score(
    y_test, y_pred, average='weighted', zero_division=0
)

print("\n----- SVM PERFORMANCE -----")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred, zero_division=0
))

# STEP 13: Confusion Matrix

labels = svm_model.named_steps['classifier'].classes_

cm = confusion_matrix(
    y_test, y_pred, labels=labels
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(cmap='Blues', xticks_rotation=45)
plt.title("SVM Confusion Matrix")
plt.tight_layout()
plt.show()

# STEP 14: Final Result

print("\nRESULT:")
print("Support Vector Machine classification completed.")
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```
